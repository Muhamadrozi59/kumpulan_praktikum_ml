```python
import pandas as pd
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, classification_report

print("Data Input dan Persiapan")

# Data Hasil Sebenarnya (True Labels)
# 'Lulus' adalah kelas Positif
y_true = ['Lulus', 'Lulus', 'Lulus', 'Lulus', 'Lulus', 
          'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus']

# Data Hasil Prediksi Model (Predicted Labels)
y_pred = ['Lulus', 'Lulus', 'Lulus', 'Tidak Lulus', 'Tidak Lulus', 
          'Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus']

# Visualisasi data awal
df_data = pd.DataFrame({
    'NIM': [f'TI00{i}' for i in range(1, 11)],
    'Hasil Sebenarnya': y_true,
    'Hasil Prediksi': y_pred
})
print(df_data)
```

```python
print("-" * 50)
print("Membuat Confusion Matrix")

# Definisikan urutan label, 'Lulus' sebagai Positif (indeks 0)
labels = ['Lulus', 'Tidak Lulus']

# Hitung Confusion Matrix
cm = confusion_matrix(y_true, y_pred, labels=labels)

# Konversi ke DataFrame untuk representasi visual yang rapi
# Baris: Actual (Sebenarnya); Kolom: Predicted (Prediksi)
cm_df = pd.DataFrame(cm, 
                     index=[f'Sebenarnya: {l}' for l in labels], 
                     columns=[f'Prediksi: {l}' for l in labels])

print(cm_df)

# Ekstrak nilai komponen Confusion Matrix (TP, FN, FP, TN)
TP = cm[0, 0] # True Positive: Sebenarnya Lulus, Prediksi Lulus
FN = cm[0, 1] # False Negative: Sebenarnya Lulus, Prediksi Tidak Lulus
FP = cm[1, 0] # False Positive: Sebenarnya Tidak Lulus, Prediksi Lulus
TN = cm[1, 1] # True Negative: Sebenarnya Tidak Lulus, Prediksi Tidak Lulus

print(f"\nKomponen Matriks (Lulus adalah Positif):")
print(f"TP (Lulus Benar): {TP}")
print(f"FN (Lulus Salah Negatif): {FN}")
print(f"FP (Tidak Lulus Salah Positif): {FP}")
print(f"TN (Tidak Lulus Benar): {TN}")
```

```python
print("-" * 50)
print("Perhitungan Metrik Performansi")

# --- 1. Accuracy (Akurasi) ---
accuracy = accuracy_score(y_true, y_pred)
print(f"1. Accuracy (Akurasi): {accuracy:.4f} ({accuracy*100:.2f}%)")

# --- 2. Precision (Presisi) ---
# Mengukur keakuratan prediksi positif ('Lulus')
precision = precision_score(y_true, y_pred, pos_label='Lulus')
print(f"2. Precision (Presisi): {precision:.4f} ({precision*100:.2f}%)")

# --- 3. Recall (Sensitivitas) ---
# Mengukur kemampuan model menangkap semua kasus positif ('Lulus')
recall = recall_score(y_true, y_pred, pos_label='Lulus')
print(f"3. Recall (Sensitivitas/Recall): {recall:.4f} ({recall*100:.2f}%)")

print("\n--- Ringkasan Classification Report ---")

# classification_report menampilkan semua metrik secara ringkas
print(classification_report(y_true, y_pred, labels=['Lulus', 'Tidak Lulus']))
```
